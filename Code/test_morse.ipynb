## API aufrufen
Lokal: `http://127.0.0.1:8000` – später in der Cloud die render.com-URL eintragen.

```python
import json, urllib.request

BASIS_URL = 'http://127.0.0.1:8000'  # später: https://<dein-service>.onrender.com

def post(pfad, daten):
    req = urllib.request.Request(BASIS_URL + pfad, data=json.dumps(daten).encode(),
                                 headers={'Content-Type': 'application/json'})
    with urllib.request.urlopen(req) as r:
        return json.loads(r.read())

print(post('/text-zu-morse', {'text': 'Hallo Welt'}))
print(post('/morse-zu-text', {'morse': '.... .- .-.. .-.. --- / .-- . .-.. -'}))
print(post('/uebersetzen', {'eingabe': '... --- ...'}))
```
